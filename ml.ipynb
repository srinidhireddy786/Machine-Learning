```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
 

 
dataset = pd.read_csv('50_Startups.csv')
print('Dataset shape:', dataset.shape)
display(dataset.head())
 
# Separate features and target
X = dataset.iloc[:, :-1].values  # all columns except last (Profit)
y = dataset.iloc[:, -1].values   # last column (Profit)
 
print('\nFeature preview (first 5 rows):')
print(X[:5])
 
# Encode categorical 'State' column (index 3)
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [3])], remainder='passthrough')
X_encoded = ct.fit_transform(X)
 
# Convert to numpy array and inspect
X = np.array(X_encoded)
print('\nAfter OneHotEncoding, feature shape:', X.shape)
print('\nFirst 5 rows after encoding:')
print(X[:5])
```

```python
# 2) Fit Model - Code
# Splitting into training and test sets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
 
print('Training samples:', X_train.shape[0])
print('Test samples:', X_test.shape[0])
 
# Train the Multiple Linear Regression model
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
 
print('Intercept:', regressor.intercept_)
print('Number of coefficients:', len(regressor.coef_))
```

```python
# 3) Predict - Code
# Predicting Test set results
y_pred = regressor.predict(X_test)
 
# Compare Actual vs Predicted
comparison = pd.DataFrame({
    'Actual Profit': y_test,
    'Predicted Profit': np.round(y_pred,2)
})
display(comparison.reset_index(drop=True))
```

```python
# 4) Visualization - Code

plt.figure(figsize=(10,6))
plt.scatter(range(len(y_test)), y_test, color='red', label='Actual Profit', alpha=0.7, s=60)
plt.scatter(range(len(y_pred)), y_pred, color='blue', label='Predicted Profit', alpha=0.7, marker='x', s=60)
plt.title('Actual vs Predicted Profit (Test Set)')
plt.xlabel('Test Sample Index')
plt.ylabel('Profit')
plt.legend()
plt.show()

residuals = y_test - y_pred
plt.figure(figsize=(10,5))
sns.histplot(residuals, kde=True)
plt.title('Residuals Distribution (Actual - Predicted)')
plt.xlabel('Residual')
plt.show()
 

plt.figure(figsize=(10,5))
plt.scatter(y_pred, residuals)
plt.axhline(y=0, color='black', linestyle='--')
plt.title('Residuals vs Predicted')
plt.xlabel('Predicted Profit')
plt.ylabel('Residual (Actual - Predicted)')
plt.show()
 
```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
 
mse = mean_squared_error(y_test, y_pred)
 
r2 = r2_score(y_test, y_pred)
 
print(f'Mean Squared Error (MSE): {mse:.2f}')
print(f'R^2 Score: {r2:.4f}')
 
raw_sample = np.array([[160000, 130000, 300000, 'California']])
encoded_sample = ct.transform(raw_sample)
encoded_sample = np.array(encoded_sample, dtype=float)
pred_sample = regressor.predict(encoded_sample)
print('\nSample startup predicted profit:', float(pred_sample[0]))
```

Higher flexibility → lower training error

Too much flexibility → overfitting

Degree 2 usually gives best bias–variance trade-off

A straight-line (linear) model assumes that car price increases at a constant rate with engine size.
However, in reality:
For small engines, price increases slowly
For medium engines, price increases rapidly
For large engines, price growth slows due to luxury saturation

Recommended Model: Polynomial Regression with a moderate degree (degree 2 or 3)

```python
X = df.drop(['cnt','dteday'], axis=1)
y = df['cnt']
```

```python
cat_cols = X.select_dtypes(include='object').columns
num_cols = X.select_dtypes(exclude='object').columns

for col in cat_cols:
    X[col] = X[col].astype(str)
    
```

```python
encoder = OneHotEncoder(drop='first', sparse_output=False)
X_cat = encoder.fit_transform(X[cat_cols])

scaler = StandardScaler()
X_num = scaler.fit_transform(X[num_cols])

X_final = np.hstack((X_num, X_cat))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_final, y, test_size=0.2, random_state=42
)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
```

```python
alphas = [0.01, 0.1, 1, 10, 100]
best_ridge_r2 = -1

for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(X_train, y_train)
    preds = ridge.predict(X_test)
    score = r2_score(y_test, preds)

    if score > best_ridge_r2:
        best_ridge_r2 = score
        best_alpha_ridge = a
        best_ridge_model = ridge

y_pred_ridge = best_ridge_model.predict(X_test)
best_alpha_ridge

```

```python
alphas = [0.001, 0.01, 0.1, 1]
best_lasso_r2 = -1

for a in alphas:
    lasso = Lasso(alpha=a, max_iter=5000)
    lasso.fit(X_train, y_train)
    preds = lasso.predict(X_test)
    score = r2_score(y_test, preds)

    if score > best_lasso_r2:
        best_lasso_r2 = score
        best_alpha_lasso = a
        best_lasso_model = lasso

y_pred_lasso = best_lasso_model.predict(X_test)
best_alpha_lasso
```

```python
def evaluate_model(y_true, y_pred):
    print("R2 Score :", r2_score(y_true, y_pred))
    print("MSE      :", mean_squared_error(y_true, y_pred))
    print("RMSE     :", np.sqrt(mean_squared_error(y_true, y_pred)))
    print("MAE      :", mean_absolute_error(y_true, y_pred))
```

```python
print("Linear Regression")
evaluate_model(y_test, y_pred_lr)

print("\nRidge Regression")
evaluate_model(y_test, y_pred_ridge)

print("\nLasso Regression")
evaluate_model(y_test, y_pred_lasso)
```
